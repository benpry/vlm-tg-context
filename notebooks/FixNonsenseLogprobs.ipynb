```python
from openai import OpenAI
from src.utils import get_logprobs_from_openai_choice
```

```python
api_base = "http://localhost:8000/v1"
client = OpenAI(base_url=api_base)
```

```python
completion = client.chat.completions.create(
    model="Qwen/Qwen2.5-VL-32B-Instruct",
    messages=[
        {
            "role": "user",
            "content": "Please respond with just one capital letter. You can choose any letter you want.",
        }
    ],
    max_tokens=1,
    temperature=1,
    logprobs=True,
    top_logprobs=1000,
)
```

```python
completion.choices[0].logprobs.content[0].top_logprobs
```

```python
for top_lp in completion.choices[0].logprobs.content[0].top_logprobs:
    if top_lp.token in [
        "A",
        "B",
        "C",
        "D",
        "E",
        "F",
        "G",
        "H",
        "I",
        "J",
        "K",
        "L",
    ]:
        print(top_lp.token, top_lp.logprob)
```
